```python
# Định nghĩa hàm để đếm số lượng giá trị ngoại lệ (outliers) trong một tập dữ liệu
def count_outliers(data):
    # Tính tứ phân vị thứ nhất (25%) - giá trị mà tại đó 25% dữ liệu nằm dưới nó
    Q1 = data.quantile(0.25)

    # Tính tứ phân vị thứ ba (75%) - giá trị mà tại đó 75% dữ liệu nằm dưới nó
    Q3 = data.quantile(0.75)

    # Tính khoảng biến thiên tứ phân vị (Interquartile Range)
    IQR = Q3 - Q1

    # Xác định ngưỡng dưới (Lower Bound) và ngưỡng trên (Upper Bound) theo quy tắc 1.5 * IQR
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    # Trả về tổng số lượng các giá trị nằm ngoài khoảng an toàn (nhỏ hơn ngưỡng dưới hoặc lớn hơn ngưỡng trên)
    return ((data < lower) | (data > upper)).sum()

# Áp dụng hàm count_outliers cho tất cả các cột có kiểu dữ liệu là số (số nguyên hoặc số thực)
outlier_counts = df.select_dtypes(include=['int64', 'float64']).apply(count_outliers)

# In ra màn hình kết quả: chỉ hiển thị những cột nào có tồn tại giá trị ngoại lệ (> 0)
print("Số lượng Outliers trong từng cột:")
print(outlier_counts[outlier_counts > 0])
```

```python
# Danh sách các biến số quan trọng có chứa giá trị ngoại lệ cần được xử lý
cols_to_fix = ['MonthlyIncome', 'TotalWorkingYears', 'YearsAtCompany', 'YearsSinceLastPromotion']

# Duyệt qua từng cột trong danh sách để thực hiện xử lý
for col in cols_to_fix:
    # Tính toán tứ phân vị thứ nhất (25%) và thứ ba (75%) của cột đó
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    # Tính khoảng biến thiên tứ phân vị (IQR)
    IQR = Q3 - Q1

    # Xác định ngưỡng trên (Upper Limit) để nhận diện các giá trị lớn bất thường
    upper_limit = Q3 + 1.5 * IQR

    # Kỹ thuật Imputation: Thay thế tất cả các giá trị vượt quá ngưỡng trên bằng chính giá trị ngưỡng đó
    # Hàm .clip(upper=...) giúp cắt ngọn các giá trị ngoại lệ mà không làm mất dòng dữ liệu nào
    df[col] = df[col].clip(upper=upper_limit)

# Thông báo sau khi hoàn tất quá trình xử lý làm sạch dữ liệu
print("Đã hoàn thành Imputation cho các cột ngoại lệ!")

# Kiểm tra lại số lượng dòng để xác nhận rằng dữ liệu vẫn được bảo toàn đầy đủ
print(f"Số lượng dòng hiện tại: {len(df)}")
```

```python
from scipy import stats
import seaborn as sns
import matplotlib.pyplot as plt

# BƯỚC 1: Kết nối dữ liệu cần thiết (Lấy Attrition từ bảng df gốc và MonthlyIncome từ df_job_details)
# Chúng ta dùng 'EmployeeNumber' làm chìa khóa kết nối
df_hypo = pd.merge(df[['EmployeeNumber', 'Attrition']],
                   df_job_details[['EmployeeNumber', 'MonthlyIncome']],
                   on='EmployeeNumber')

# BƯỚC 2: Tách dữ liệu thành 2 nhóm để so sánh
salary_stay = df_hypo[df_hypo['Attrition'] == 'No']['MonthlyIncome']
salary_leave = df_hypo[df_hypo['Attrition'] == 'Yes']['MonthlyIncome']

# BƯỚC 3: Thực hiện kiểm định T-test (Independent Samples T-test)
t_stat, p_value = stats.ttest_ind(salary_stay, salary_leave)

# BƯỚC 4: Xuất kết quả
print(f"--- KẾT QUẢ KIỂM ĐỊNH T-TEST ---")
print(f"Lương trung bình nhóm Ở LẠI: {salary_stay.mean():.2f}")
print(f"Lương trung bình nhóm NGHỈ VIỆC: {salary_leave.mean():.2f}")
print(f"P-value: {p_value}")

# BƯỚC 5: Kết luận về mặt thống kê
alpha = 0.05
if p_value < alpha:
    print("\n=> KẾT LUẬN: Bác bỏ giả thuyết H0. Có sự khác biệt đáng kể về lương giữa 2 nhóm.")
    print("Người nghỉ việc có mức lương trung bình thấp hơn rõ rệt.")
else:
    print("\n=> KẾT LUẬN: Chưa đủ bằng chứng để bác bỏ H0. Lương không phải yếu tố khác biệt chính.")
```

```python
from scipy import stats

# BƯỚC 1: Kết nối dữ liệu (Lấy Attrition từ df gốc và YearsWithCurrManager từ df_work_history)
df_hypo3 = pd.merge(df[['EmployeeNumber', 'Attrition']],
                    df_work_history[['EmployeeNumber', 'YearsWithCurrManager']],
                    on='EmployeeNumber')

# BƯỚC 2: Tách nhóm
manager_years_stay = df_hypo3[df_hypo3['Attrition'] == 'No']['YearsWithCurrManager']
manager_years_leave = df_hypo3[df_hypo3['Attrition'] == 'Yes']['YearsWithCurrManager']

# BƯỚC 3: Thực hiện kiểm định T-test
t_stat, p_value = stats.ttest_ind(manager_years_stay, manager_years_leave)

# BƯỚC 4: Xuất kết quả thẳng thắn
print(f"--- KIỂM ĐỊNH MỐI QUAN HỆ VỚI QUẢN LÝ ---")
print(f"Số năm trung bình làm với sếp (Người ở lại): {manager_years_stay.mean():.2f} năm")
print(f"Số năm trung bình làm với sếp (Người nghỉ việc): {manager_years_leave.mean():.2f} năm")
print(f"P-value: {p_value}")

# BƯỚC 5: Kết luận
alpha = 0.05
if p_value < alpha:
    print("\n=> KẾT LUẬN: Bác bỏ H0. Số năm làm việc với quản lý có ảnh hưởng đến quyết định nghỉ việc.")
    print("Nhân viên có xu hướng nghỉ việc khi thời gian làm việc với quản lý hiện tại còn ngắn.")
else:
    print("\n=> KẾT LUẬN: Chưa đủ bằng chứng để nói rằng mối quan hệ với quản lý gây nghỉ việc.")
```

```python
from scipy import stats
import seaborn as sns
import matplotlib.pyplot as plt

# BƯỚC 1: Kết nối dữ liệu (Lấy Attrition và DistanceFromHome)
# Giả sử DistanceFromHome nằm trong bảng df gốc của bạn
df_hypo_dist = df[['EmployeeNumber', 'Attrition', 'DistanceFromHome']]
# BƯỚC 2: Tách dữ liệu thành 2 nhóm để so sánh khoảng cách
dist_stay = df_hypo_dist[df_hypo_dist['Attrition'] == 'No']['DistanceFromHome']
dist_leave = df_hypo_dist[df_hypo_dist['Attrition'] == 'Yes']['DistanceFromHome']
# BƯỚC 3: Thực hiện kiểm định T-test (Independent Samples T-test)
# Kiểm định xem trung bình khoảng cách của 2 nhóm có bằng nhau không
t_stat, p_value = stats.ttest_ind(dist_stay, dist_leave)
# BƯỚC 4: Xuất kết quả thống kê mô tả
print(f"--- KẾT QUẢ KIỂM ĐỊNH T-TEST CHO KHOẢNG CÁCH ---")
print(f"Khoảng cách TB nhóm Ở LẠI: {dist_stay.mean():.2f} km")
print(f"Khoảng cách TB nhóm NGHỈ VIỆC: {dist_leave.mean():.2f} km")
print(f"P-value: {p_value}")
# BƯỚC 5: Kết luận dựa trên mức ý nghĩa alpha = 0.05
alpha = 0.05
if p_value < alpha:
    print("\n=> KẾT LUẬN: Bác bỏ giả thuyết H0. Có sự khác biệt đáng kể về khoảng cách giữa 2 nhóm.")
    print("Khoảng cách di làm xa là một yếu tố tác động thực sự đến quyết định nghỉ việc.")
else:
    print("\n=> KẾT LUẬN: Chưa đủ bằng chứng để bác bỏ H0. Khoảng cách không phải là yếu tố tạo ra sự khác biệt rõ rệt.")
```
